1. Select transaction_id, customer_id, and amount

```python
df.select("transaction_id","customer_id","amount").show()
```

2. Filter transactions above ₹50,000

```python
df.filter(F.col("amount") > 50000).show()
```

3. Filter only DEBIT transactions

```python
df.filter(F.col("transaction_type") == "DEBIT").show()
```

4. Create transaction_fee as 1% of amount

```python
df.withColumn("transaction_fee",F.col("amount") * 0.01).show()
```

5. Create net_amount after deducting the fee

```python
df.withColumn(
    "transaction_fee",
    F.col("amount") * 0.01
).withColumn(
    "net_amount",
    F.col("amount") - F.col("transaction_fee")
).show()
```

6. Rename city to branch_city

```python
df.withColumnRenamed("city","branch_city").show()
```

7. Drop an unnecessary column

```python
df.drop("city").show()
```

8. Find distinct account types

```python
df.select("account_type").distinct().show()
```

9. Remove duplicate transaction IDs

```python
df.dropDuplicates(["transaction_id"]).show()
```

10. Sort transactions by amount descending

```python
df.orderBy(F.col("amount").desc()).show()
```

11. Group by account type and calculate count, total, average, min, and max

```python
df.groupBy("account_type").agg(
    F.count("*").alias("count"),
    F.sum("amount").alias("total"),
    F.avg("amount").alias("average"),
    F.min("amount").alias("min"),
    F.max("amount").alias("max")
).show()
```

12. Create risk_level

Rules:

HIGH → amount >= ₹100,000
MEDIUM → amount >= ₹50,000
LOW → otherwise

```python
df.withColumn(
    "risk_level",
    F.when(F.col("amount") >= 100000, "HIGH")
    .when(F.col("amount") >= 50000, "MEDIUM")
    .otherwise("LOW")
).show()
```

13. Convert date to DateType and extract year, month, and day

```python
df.withColumn("transaction_date",F.to_date(F.col("transaction_date"))).select(
    F.year("transaction_date").alias("year"),
    F.month("transaction_date").alias("month"),
    F.dayofmonth("transaction_date").alias("day")
).show()
```

14.Handle null values in amount and city.

```python
df.fillna({"amount":0.0,
           "city":"unknown"}).show()
```

```python
customer_data = [
    ("C101", "Rahul", "Bengaluru"),
    ("C102", "Priya", "Hyderabad"),
    ("C103", "Amit", "Chennai"),
    ("C104", "Sneha", "Pune"),
    ("C105", "Kiran", "Bengaluru"),
    ("C106", "Arjun", "Hyderabad"),
    ("C107", "Neha", "Chennai"),
    ("C108", "Ravi", "Bengaluru"),
    ("C109", "Anjali", "Pune"),
    ("C110", "Vikram", "Bengaluru")
]
customer_df = spark.createDataFrame(
    customer_data,
    ["customer_id", "customer_name", "customer_city"]
)
customer_df.show()

##joins
joined_df = df.join(customer_df,"customer_id","left")

joined_df.show()
```

17. Repartition by account_type and explain why

```python
repart_df = df.repartition(4,"account_type")

print("Number of partitions:",repart_df.rdd.getNumPartitions())
##repartition() distributes the data based on account_type.
## It causes a shuffle and can be useful when subsequent processing is based on account_type, such as groupBy().
```

8. Build a final banking transformation pipeline producing clean and enreached transactions

```python
final_df = (    ##droping duplicants
    df
    .dropDuplicates(["transaction_id"])
    ##filling null values
    .fillna({
        "amount": 0.0,
        "city": "Unknown"
    })
    ##date functions
    .withColumn(
        "transaction_date",
        F.to_date("transaction_date")
    )
    .withColumn(
        "transaction_fee",
        F.round(F.col("amount") * 0.01, 2)
    )
    .withColumn(
        "net_amount",
        F.round(
            F.col("amount") - F.col("transaction_fee"),
            2
        )
    )
    .withColumn(              #creating new column with assigning values based on performances
        "risk_level",
        F.when(F.col("amount") >= 100000, "HIGH")
         .when(F.col("amount") >= 50000, "MEDIUM")
         .otherwise("LOW")
    )
    .select(       #selecting all columns
        "transaction_id",
        "customer_id",
        "account_type",
        "transaction_type",
        "amount",
        "transaction_fee",
        "net_amount",
        "city",
        "transaction_date",
        "risk_level"
    )
)

final_df.show(truncate=False)
```
