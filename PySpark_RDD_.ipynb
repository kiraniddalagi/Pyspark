```python
#Q1 Dataset [5, 10, 15, 20, 25, 30] Problem: Create an RDD from the dataset and multiply every number by 2.
from pyspark import SparkContext

sc = SparkContext.getOrCreate()
data = [5, 10, 15, 20, 25, 30]
rdd = sc.parallelize(data)
result = rdd.map(lambda x: x * 2).collect()
print(result)
```

```python
#Q23 Calculate the frequency of every word.
from pyspark import SparkContext
sc = SparkContext.getOrCreate()
data = ["spark is fast", "spark is powerful", "spark is easy", "hadoop is distributed", "spark is fast"]
rdd = sc.parallelize(data)
word_counts = rdd.flatMap(lambda x: x.split()).map(lambda x: (x, 1)).reduceByKey(lambda x, y: x + y).collect()
for word, count in word_counts:
    print(f"{word}: {count}")
```

```python
#Q24 Problem: Find the most frequently occurring word.
from pyspark.sql import SparkSession
spark = SparkSession.builder.getOrCreate()
dataset =[
    "spark is fast",
    "spark is powerful",
    "spark is easy",
    "hadoop is distributed",
    "spark is fast"
]
sc = spark.sparkContext
rdd = sc.parallelize(dataset)
rdd2 = rdd.flatMap(lambda x: x.split(" "))
rdd3 = rdd2.map(lambda x: (x,1))
rdd4 = rdd3.reduceByKey(lambda x,y: x+y)
rdd5 = rdd4.sortBy(lambda x: x[1], ascending=False)
print(rdd5.take(1))
```

```python
#Q25 Problem: Find all words that occur more than once.
from pyspark import SparkContext
sc = SparkContext.getOrCreate()
rdd = sc.parallelize(dataset)
rdd2 = rdd.flatMap(lambda x: x.split(" "))
rdd3 = rdd2.map(lambda x: (x,1))
rdd4 = rdd3.reduceByKey(lambda x,y: x+y)
rdd5 = rdd4.filter(lambda x: x[1] > 1)
print(rdd5.collect())
for word, count in rdd5.collect():
    print(f"{word}: {count}")
```

```python
#Q33 Problem: Calculate total revenue for every product.

from pyspark.sql import SparkSession
spark = SparkSession.builder.getOrCreate()
Dataset  = [
 ("Laptop", 2, 50000),
 ("Mobile", 5, 20000),
 ("Laptop", 1, 50000),
 ("Tablet", 3, 15000),
 ("Mobile", 2, 20000),
 ("Laptop", 4, 50000),
 ("Tablet", 2, 15000),
 ("Mobile", 3, 20000)
]
sc = spark.sparkContext
rdd = sc.parallelize(Dataset)
rdd2 = rdd.map(lambda x: (x[0], x[1]*x[2]))
rdd3 = rdd2.reduceByKey(lambda x,y: x+y)
result = rdd3.collect()
for product, revenue in result:
    print(f"{product} -> {revenue}")
```
