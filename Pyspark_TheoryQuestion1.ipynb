9.write difference of files CSV,AVRO, parquet, ORC

1. CSV

CSV = Comma-Separated Values

Simple text format.
Human readable.
Easy to exchange between applications.
Does not normally contain schema or data types.
Larger storage size.
Slower for large-scale analytics.

Best use: Source files, simple data exchange, small datasets.

2. Avro

Avro is a row-oriented binary format.

Stores schema along with data.
Supports schema evolution.
Good for serialization/deserialization.
Commonly used with streaming systems such as Kafka.
Suitable when entire records need to be processed.

Best use: Data serialization, Kafka/streaming, record-oriented data.

3. Parquet

Parquet is a columnar storage format.

Stores data by columns.
Excellent compression.
Supports schema.
Efficient for analytical queries.
Supports column pruning and predicate pushdown.
Very commonly used with Spark and data lakes.

Best use: Big Data analytics, Spark, Data Lakes, ETL/ELT.

4. ORC

ORC = Optimized Row Columnar

Despite the name, ORC is primarily a columnar storage format.

High compression.
Efficient analytical queries.
Stores schema and metadata.
Supports predicate pushdown.
Commonly associated with Hive and Hadoop ecosystems.

Best use: Hive-based and large-scale analytical workloads.

ANS:: Spark SQL

Spark SQL is a Spark module used to process and analyze structured and semi-structured data using SQL queries. It allows users to create temporary views or tables and execute SQL operations such as SELECT, JOIN, GROUP BY, WHERE, and ORDER BY. Spark SQL uses Spark's Catalyst Optimizer and Tungsten execution engine to optimize query execution

DataFrame API

The DataFrame API is a programming interface in Spark used to work with structured data in the form of rows and columns. It provides built-in functions and methods such as select(), filter(), withColumn(), groupBy(), and join() for data transformation and analysis. DataFrames are distributed and immutable, and Spark optimizes their execution using the Catalyst Optimizer.

15.how do you create a table In dataframe

We use createOrReplaceTempView() when we want to register a DataFrame as a temporary SQL view for the current Spark session. We use saveAsTable() when we want to persist the DataFrame as a table that can be accessed through the Spark catalog.

```python
from pyspark.sql import SparkSession

# Create Spark Session
spark = SparkSession.builder \
    .appName("TemporaryViewExample") \
    .getOrCreate()

# Sample data
data = [
    (101, "Rahul", "IT", 50000),
    (102, "Anita", "HR", 60000),
    (103, "Kiran", "Finance", 55000),
    (104, "Arjun", "IT", 70000)
]

# Create DataFrame
df = spark.createDataFrame(
    data,
    ["employee_id", "name", "department", "salary"]
)

# Display DataFrame
df.show()

# Create temporary view
df.createOrReplaceTempView("employee")

# Query temporary view using Spark SQL
result = spark.sql("""
    SELECT employee_id, name, department, salary
    FROM employee
    WHERE salary > 55000
""")

result.show()
```
